```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. 设置全局学术字体与样式 ---
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial', 'Helvetica']  # 确保英文和符号正常显示
plt.rcParams['axes.unicode_minus'] = False  # 正常显示负号

# --- 2. 准备数据 ---
# 表 7 ED 任务数据
ed_labels = ['Ours', 'w/o MacBERT', 'w/o ELAM']
ed_p = [85.23, 84.02, 70.54]
ed_r = [80.66, 79.59, 78.75]
ed_f1 = [82.88, 81.74, 74.42]

# 表 8 EAE 任务数据
eae_labels = ['Ours', 'w/o MacBERT', 'w/o MMoE', 'w/o MCAM']
eae_p = [89.12, 72.81, 89.07, 69.25]
eae_r = [67.44, 66.18, 56.30, 64.37]
eae_f1 = [76.78, 69.34, 68.99, 66.72]

# --- 3. 开始画图 (1行2列组合图) ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 5.5), dpi=300)

# 定义高级学术色系
colors = {
    'P': '#1f77b4',    # 沉稳深蓝
    'R': '#e377c2',    # 优雅粉红
    'F1': '#2ca02c'    # 质感复古绿
}

# ----------------- 左图：ED 任务 -----------------
# 画折线
ax1.plot(ed_labels, ed_p, marker='o', ls='-', lw=2, color=colors['P'], label='P', ms=7)
ax1.plot(ed_labels, ed_r, marker='s', ls='--', lw=2, color=colors['R'], label='R', ms=7)
ax1.plot(ed_labels, ed_f1, marker='^', ls='-.', lw=2.5, color=colors['F1'], label='F1', ms=8)

# 标注具体数值 (可选，为了让图更清晰)
for i, (p, r, f1) in enumerate(zip(ed_p, ed_r, ed_f1)):
    ax1.text(i, p + 0.8, f"{p:.1f}", ha='center', va='bottom', fontsize=9, color=colors['P'])
    ax1.text(i, r - 1.5, f"{r:.1f}", ha='center', va='top', fontsize=9, color=colors['R'])
    ax1.text(i, f1 + 0.8, f"{f1:.1f}", ha='center', va='bottom', fontsize=9, color=colors['F1'])

ax1.set_title('(a) ED Task Ablation', fontsize=13, fontweight='bold', pad=15)
ax1.set_ylabel('Performance (%)', fontsize=11, fontweight='bold')
ax1.set_ylim(65, 95)  # 适当留白

# ----------------- 右图：EAE 任务 -----------------
# 画折线
ax2.plot(eae_labels, eae_p, marker='o', ls='-', lw=2, color=colors['P'], label='P', ms=7)
ax2.plot(eae_labels, eae_r, marker='s', ls='--', lw=2, color=colors['R'], label='R', ms=7)
ax2.plot(eae_labels, eae_f1, marker='^', ls='-.', lw=2.5, color=colors['F1'], label='F1', ms=8)

# 标注具体数值
for i, (p, r, f1) in enumerate(zip(eae_p, eae_r, eae_f1)):
    ax2.text(i, p + 0.8, f"{p:.1f}", ha='center', va='bottom', fontsize=9, color=colors['P'])
    ax2.text(i, r - 1.5, f"{r:.1f}", ha='center', va='top', fontsize=9, color=colors['R'])
    ax2.text(i, f1 + 0.8, f"{f1:.1f}", ha='center', va='bottom', fontsize=9, color=colors['F1'])

ax2.set_title('(b) EAE Task Ablation', fontsize=13, fontweight='bold', pad=15)
ax2.set_ylim(50, 95)  # EAE波动较大，调整y轴范围保持美观

# --- 4. 统一的细节打磨 (网格、边框、图例) ---
for ax in [ax1, ax2]:
    # 开启轻微的横向网格线
    ax.grid(axis='y', linestyle=':', alpha=0.6, color='#cccccc')
    
    # 去掉上方和右方的无用边框，更显高级
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_color('#333333')
    ax.spines['bottom'].set_color('#333333')
    
    # 调整刻度字体大小
    ax.tick_params(axis='both', labelsize=10)

# 合并图例，放在整个大图的上方中央
handles, labels = ax1.get_legend_handles_labels()
fig.legend(handles, labels, loc='upper center', bbox_to_anchor=(0.5, 0.96), 
           ncol=3, frameon=True, facecolor='white', edgecolor='none', fontsize=11)

# 调整整体布局，防止标签重叠
plt.tight_layout(rect=[0, 0, 1, 0.90])

# --- 5. 保存高质量图片 ---
# pdf 格式适合 LaTeX 论文无损放大，png 格式方便 Word 插入
plt.savefig('ablation_results.pdf', bbox_inches='tight')
plt.savefig('ablation_results.png', dpi=900, bbox_inches='tight')

print("高级感消融实验折线图已成功生成并保存！")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. 数据准备 ---
tasks = ['Task 1', 'Task 2', 'Task 3']
experts = [
    'Expert 1 (E1): Local Position', 
    'Expert 2 (E2): Long-dist Struct', 
    'Expert 3 (E3): Global Semantic', 
    'Expert 4 (E4): General Aux'
]

data = [
    [0.48, 0.15, 0.18], # E1
    [0.12, 0.42, 0.15], # E2
    [0.15, 0.18, 0.45], # E3
    [0.25, 0.25, 0.22]  # E4
]

# --- 2. 风格与配色设置 ---
# 设置为纯白背景样式
plt.style.use('default') 
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# 清新蓝色系调色盘 (从深蓝到清新绿蓝)
colors = ['#1a5e9a', '#3f8fca', '#70adcf', '#a5d8dd'] 

x = np.arange(len(tasks))
width = 0.18

fig, ax = plt.subplots(figsize=(10, 6), dpi=300)

# --- 3. 绘制柱状图 ---
for i in range(len(data)):
    # 移除黑色边框，使用更淡的同色系深色或者直接无边框
    rects = ax.bar(x + (i - 1.5) * width, data[i], width, 
                   label=experts[i], color=colors[i], 
                   edgecolor='white', linewidth=1, zorder=3)
    
    # 标注数值
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 5), 
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=9, 
                    color='#333333', fontweight='500')

# --- 4. 极致简约美化 ---
ax.set_ylabel('Weight Score', fontsize=12, color='#444444')
ax.set_xticks(x)
ax.set_xticklabels(tasks, fontsize=12, color='#222222')

# 仅保留极细的 y 轴网格线
ax.yaxis.grid(True, linestyle='-', alpha=0.2, color='gray', zorder=0)

# 隐藏顶部和右侧边框
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)
ax.spines['left'].set_color('#999999')
ax.spines['bottom'].set_color('#999999')

# 限制 y 轴范围，留出上方空间
ax.set_ylim(0, 0.6)

# --- 5. 图例优化 ---
# 放在下方，取消边框，横向排列
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.12), 
          ncol=2, frameon=False, fontsize=10, columnspacing=1.5)

plt.tight_layout()

# 保存
plt.savefig('mmoe_blue_clean.png', dpi=900, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. 数据准备与预处理 ---
labels = ['EAE F1', 'Efficiency\n(Low Latency)', 'Cost-Effective\n(No Token)', 'Deployability']
num_vars = len(labels)

# 数据量化处理 (均转化成 0-1 之间的得分，越高代表越好)
# 顺序：LLM Zero, LLM Few, LLM LoRA, Ours
# F1 分数归一化 (直接用原始百分比/100)
f1_scores = [0.523, 0.617, 0.729, 0.7745]
# 延迟得分 (延迟越低分越高)
latency_scores = [0.1, 0.05, 0.15, 0.95] 
# 成本得分 (Token消耗越少分越高，Ours不需要Token定为满分)
cost_scores = [0.4, 0.2, 0.45, 1.0]
# 部署性 (低:0.3, 中:0.6, 高:1.0)
deploy_scores = [0.6, 0.6, 0.3, 1.0]

data = [
    [f1_scores[i], latency_scores[i], cost_scores[i], deploy_scores[i]] 
    for i in range(4)
]

# 闭合雷达图数据
plot_data = [np.concatenate((d, [d[0]])) for d in data]
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]

# --- 2. 绘图设置 ---
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True), dpi=300)
ax.set_facecolor('white')

# 设置配色：3种 LLM 用渐变灰/浅色，Ours 用标志性的清新蓝
colors = ['#CCCCCC', '#999999', '#666666', '#1a5e9a'] 
line_styles = ['--', '--', '--', '-']
methods = ['LLM Zero-shot', 'LLM Few-shot', 'LLM LoRA', 'Ours (Full)']

# --- 3. 绘制多边形 ---
for i in range(len(methods)):
    ax.plot(angles, plot_data[i], color=colors[i], linewidth=2, 
            linestyle=line_styles[i], label=methods[i], marker='o', markersize=5)
    # 填充 Ours 使得对比更明显
    if i == 3:
        ax.fill(angles, plot_data[i], color=colors[i], alpha=0.15)

# --- 4. 细节美化 ---
# 设置刻度标签
ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels, fontsize=11, fontweight='bold', color='#333333')

# 隐藏极坐标的数值刻度
ax.set_yticklabels([])
ax.spines['polar'].set_visible(True)
ax.spines['polar'].set_color('#EEEEEE')

# 增加背景同心圆网格
ax.grid(color='#EEEEEE', linestyle='-', linewidth=1)

# --- 5. 图例与保存 ---
plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1.1), frameon=False, fontsize=10)
plt.title('Comprehensive Comparison: Ours vs. LLMs', pad=30, fontsize=15, fontweight='bold')

plt.tight_layout()
plt.savefig('comparison_radar.png', bbox_inches='tight')
plt.show()
```
